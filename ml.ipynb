# 2. K-Means

```python
import numpy as np
from typing import List, Tuple

class Solution:
    def kMeans(
        self,
        n: int,
        k: int,
        max_iter: int,
        nums: List[List[float]]
    ) -> Tuple[List[List[float]], List[int]]:

        points = np.array(nums, dtype=float)

        centers = points[:k].copy()

        labels = np.full(n, -1, dtype=int)

        for _ in range(max_iter):
            dists = np.sum(
                (points[:, None, :] - centers[None, :, :]) ** 2,
                axis=2
            )

            new_labels = np.argmin(dists, axis=1)

            if np.array_equal(labels, new_labels):
                break

            labels = new_labels

            for j in range(k):
                cluster_points = points[labels == j]
                if len(cluster_points) > 0:
                    centers[j] = np.mean(cluster_points, axis=0)

        centers_ret = [
            [round(float(x), 4), round(float(y), 4)]
            for x, y in centers
        ]

        return centers_ret, labels.tolist()

n = 4
k = 2
max_iter = 100
nums = [[1.0,1.0],[1.5,2.0],[5.0,8.0],[8.0, 8.0]]
centers, labels = Solution().kMeans(n, k, max_iter, nums)
print("Centers:", centers)
print("Labels:", labels)
```
