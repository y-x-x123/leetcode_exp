```python
nums = [9,7,8,6,6]
n = 5
target_sum = 22
nums.sort(reverse=True)
num_people = (n//2 if n%2==0 else n//2+1)
count = 0
def backtrack(index,start,cur_sum):
    if index == num_people:
        if cur_sum >= target_sum:
            global count
            count += 1
        return
    for i in range(start,n):
        cur_sum += nums[i]
        backtrack(index+1,i+1,cur_sum)
        cur_sum -= nums[i]
backtrack(0,0,0)
print(count)
```

```python
nums = [9, 7, 8, 6, 6]
n = 5
target_sum = 22

nums.sort(reverse=True)

k = (n + 1) // 2
count = 0


def backtrack(start, selected, cur_sum):
    global count

    if selected == k:
        if cur_sum >= target_sum:
            count += 1
        return

    for i in range(start, n):
        backtrack(
            i + 1,
            selected + 1,
            cur_sum + nums[i]
        )

backtrack(0, 0, 0)

print(count)
```
